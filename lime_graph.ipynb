```python
from PIL import Image

images = [
    Image.open("lime_adaboost_sylhet.png"),
    Image.open("lime_xgboost_sylhet.png"),
    Image.open("lime_lightgbm_sylhet.png"),
    Image.open("lime_catboost_sylhet.png")
]

width, height = images[0].size
images = [img.resize((width, height)) for img in images]

combined = Image.new("RGB", (width * 2, height * 2), "white")

combined.paste(images[0], (0, 0))          # AdaBoost
combined.paste(images[1], (width, 0))      # XGBoost
combined.paste(images[2], (0, height))     # LightGBM
combined.paste(images[3], (width, height)) # CatBoost

combined.save("sdd_4_model.png", dpi=(300, 300))

print("Selesai!")
```
